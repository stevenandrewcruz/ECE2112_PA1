Problem A: Word Rotation Problem 

```python
def rotate_word(text):        # Slice from the second charter to the last, and affix the first character        
    return text[1:] + text[0]
```

```python
print(rotate_word("python"))  # Output: "ythonp"
```

```python
print(rotate_word("logic"))   # Output: "ogicl"
```

```python
print(rotate_word("Code"))    # Output: "odeC"
```

```python
print(rotate_word("A"))       # Output: "A"
```

Problem B: Username Builder Problem 

```python
def make_username(first_name, last_name):
    clean_first = first_name.lower().replace(" ", "")
    clean_last = last_name.lower().replace(" ", "")
    return f"{clean_first}.{clean_last}"

    # Lowercase and remove spaces from both names, and then join with a period.
```

```python
print(make_username("Ada" , "Lovelace"))  # Output: "ada.lovelace"
```

```python
print(make_username("Alan" , "Turing"))   # Output: "alan.turing"
```

```python
print(make_username("Ana Maria" , "De Leon"))  # Output: "anamaria.deleon"
```

Problem C: Bookened Swap Problem 

```python
def swap_bookends(items):
    # Extend Sequence Unpacking 
    first, *middle, last = items 
    # Return a new list with first and last swapped, while keeping middle in place 
    return [last] + middle + [first]
```

```python
print (swap_bookends([1,2,3,4,5,6]))  # Output: "[6,2,3,4,5,1]"
```

```python
print (swap_bookends(["red","green","blue"]))  # Output: ["blue", "green", "red"]
```

```python
print (swap_bookends([8,3]))  # Output: "[3,8]"
```
